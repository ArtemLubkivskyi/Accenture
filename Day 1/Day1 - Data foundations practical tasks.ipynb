### 8. Identify potential outliers in `alcohol` using the IQR rule. Flag values below `Q1 - 1.5 × IQR` or above `Q3 + 1.5 × IQR`.

```python
Q1 = wine_data['alcohol'].quantile(0.25)
Q3 = wine_data['alcohol'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Lower bound = {lower_bound:.2f}, Upper bound = {upper_bound:.2f}")

outliers = wine_data[(wine_data['alcohol'] < lower_bound) | (wine_data['alcohol'] > upper_bound)]
outliers_count = outliers.shape[0]

print(f"Number of outliers in 'alcohol': {outliers_count}")
outliers[['alcohol', 'quality']]
```

### 11. Plot a scatter plot of `quality` against the feature with the strongest correlation to it. Does alcohol content appear associated with wine quality?

```python
plt.figure(figsize=(10, 6))
plt.scatter(wine_data['alcohol'], wine_data['quality'], alpha=0.3, color='steelblue')
plt.title('Wine Quality vs Alcohol Content', fontsize=14, fontweight='bold')
plt.xlabel('Alcohol (%)')
plt.ylabel('Quality')
plt.grid(linestyle='--', alpha=0.5)
avg_alcohol = wine_data.groupby('quality')['alcohol'].mean()
plt.plot(avg_alcohol.values, avg_alcohol.index, color='crimson', marker='o', linewidth=2, label='Average alcohol')
plt.legend()
plt.show()
```

**Answer:** Yes, alcohol is the feature most correlated with quality (0.48). Average alcohol stays around 10% for quality 3–5, then climbs to about 12% for quality 8. Still, every quality level has a wide spread of alcohol values, so it's a clear trend but not a strict rule.

### 13. Create X, y for independent variables (all except `quality`) and target variable `quality`, perform k-fold k=3 split on them and calculate and print out the length of training set and testing set for each fold

```python
X = wine_data.drop(columns='quality')
y = wine_data['quality']

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import KFold

kf = KFold(n_splits=3, shuffle=True, random_state=42)

for fold, (train_index, test_index) in enumerate(kf.split(X), start=1):
    print(f"Fold {fold}: training set = {len(train_index)}, testing set = {len(test_index)}")
```

### 14. Create two new features:
- Sulfur dioxide ratio (`free sulfur dioxide` / `total sulfur dioxide`)
- Total acidity (`fixed acidity` + `volatile acidity` + `citric acid`)

```python
wine_features = wine_data.copy()

wine_features['sulfur_dioxide_ratio'] = wine_features['free sulfur dioxide'] / wine_features['total sulfur dioxide']
wine_features['total_acidity'] = wine_features['fixed acidity'] + wine_features['volatile acidity'] + wine_features['citric acid']

wine_features[['free sulfur dioxide', 'total sulfur dioxide', 'sulfur_dioxide_ratio',
               'fixed acidity', 'volatile acidity', 'citric acid', 'total_acidity']].head()
```

#### Extra feature: alcohol level
I just liked this one. I split wines into low / medium / high alcohol groups, which is simpler to read than raw percentages. Checking the average quality in each group shows the same trend as in Task 11: the stronger the wine, the higher its average score.


```python
wine_features['alcohol_level'] = pd.cut(wine_features['alcohol'],
                                        bins=[0, 10, 12, 15],
                                        labels=['low', 'medium', 'high'])

wine_features.groupby('alcohol_level', observed=True)['quality'].mean().round(2)

```

### 15. Create a sample dataset with dates for a whole week. Extract day of the week with `dayofweek` function (0=Monday, 6=Sunday). Additionally flag weekends by creating another new `is_weekend` feature.

```python
dates_df = pd.DataFrame({
    'date': pd.date_range(start='2026-11-02', periods=7, freq='D')
})

dates_df['day_of_week'] = dates_df['date'].dt.day_of_week  # dayofweek is deprecated in pandas 4
dates_df['day_name'] = dates_df['date'].dt.day_name()
dates_df['is_weekend'] = (dates_df['day_of_week'] >= 5).astype(int)

dates_df
```
