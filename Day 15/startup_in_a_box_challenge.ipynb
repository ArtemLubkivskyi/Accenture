# 🚀 Day 15 Challenge: Startup in a Box

> *"Every great company started as a crazy idea on a napkin. Your job: build the machine that turns napkins into Demo Day pitches."*

## 📖 The Story

**Moonshot Ventures**, a (fictional) venture studio, receives hundreds of wild product ideas every week. Their analysts can't keep up, so they've hired **you** as their founding AI engineer.

Your mission is to build an **AI venture studio**: a multi-agent system that takes **one wild idea** and turns it into a complete **🚀 Launch Kit**. The kit covers brand, market research, product plan, unit economics, a marketing campaign and an investor pitch, and it has to survive a tough **investor panel** at Demo Day.

You choose the framework. You design the team. You decide how the agents work together. 🧠

## 🎯 Rules of the Game

### ✅ Your system must…
1. **Use one agent framework** from Days 14–15 as its backbone: `smolagents`, `LangGraph`, `LlamaIndex`, `Google ADK` or `CrewAI`.
2. **Have at least 3 agents** with clearly different roles.
3. **Combine at least 2 agentic patterns** of your choice, and explain why in your design doc.
4. **Let agents call at least 2 tools**, from the Startup Toolkit or your own. The LLM should decide when to call them; don't call them yourself.
5. **Produce a `launch_kit` dictionary** that passes the Mission Check (section 7).
6. **Run top to bottom** after *Restart Kernel → Run All*.

### 🎨 You are free to…
- pick an idea from the deck **or invent your own**,
- design any architecture: a pipeline, a graph, a hierarchy, a debate, a swarm…,
- choose the agents, roles, prompts, tools and patterns, and how many iterations they get,
- add extra sections to the launch kit (logo prompt, landing page, customer interviews, risk register…).

### 🤝 Fair play
- The agents must generate the content. No hard-coding the answers!
- Keep a full run under **~50 LLM calls**, because we all share one Azure OpenAI quota.
- Never print, hard-code or commit API keys.

## 📦 The Launch Kit (your deliverable)

Your system must produce a Python `dict` called **`launch_kit`** with the keys below. Values can be strings, lists, dicts or Pydantic models; the format inside each section is up to you.

| Key | What it should contain |
|---|---|
| `startup_name` | A catchy name you've checked is available |
| `one_liner` | A one-sentence elevator pitch |
| `target_customers` | Who exactly it's for (personas/segments) |
| `market_insights` | Market size, growth, trends and risks |
| `competitors` | Main competitors and how you'll beat them |
| `mvp_features` | A prioritised list of MVP features |
| `pricing` | Pricing model and price points |
| `unit_economics` | LTV, CAC, LTV/CAC and payback, calculated with the tool |
| `marketing_campaign` | Slogan, launch channels and a few social posts |
| `investor_pitch` | Your Demo Day pitch (e.g. a 6-slide outline or a 200-word script) |
| `investor_feedback` | What the investor panel said (scores, verdicts, what changed) |

Extra keys are welcome. `show_launch_kit()` renders them too, and it will even display a `landing_page_html` section as a web page. 🌐

import litellm
from google.adk.agents import LlmAgent, LoopAgent, ParallelAgent, SequentialAgent

**Useful:** the workflow agents `SequentialAgent`, `ParallelAgent` and `LoopAgent`. `output_key` saves an agent's answer in session state, and `{market_report}` in another agent's instruction reads it back. `sub_agents=[...]` lets the LLM delegate.

<summary><b>👥 CrewAI</b> (use a fresh environment)</summary>

```python
from crewai import LLM, Agent, Crew, Process, Task
from crewai.tools import tool

llm = LLM(model=f"openai/{AZURE_OPENAI_DEPLOYMENT}", base_url=AZURE_OPENAI_ENDPOINT,
          api_key=os.environ["AZURE_OPENAI_API_KEY"])
market_tool = tool("Market Data")(get_market_data)

analyst = Agent(role="Market Analyst", goal="Size up startup markets", backstory="You love data.",
                tools=[market_tool], llm=llm)
task = Task(description="Summarise the {sector} market.", expected_output="3 bullet points", agent=analyst)
crew = Crew(agents=[analyst], tasks=[task], process=Process.sequential)
result = await crew.kickoff_async(inputs={"sector": "pets"})  # in notebooks, always use kickoff_async
print(result)
```

**Useful:** `context=[...]` chains tasks, `async_execution=True` runs tasks in parallel, `Process.hierarchical` with `manager_llm=llm` adds a manager agent, and `output_pydantic=MyPydanticModel` gives structured outputs.
</details>

````python
# 🧰 The Startup Toolkit: fictional mock data + framework-agnostic tools.
# Every tool is a plain Python function with type hints and a docstring,
# so you can wrap it with any framework's tool helper (see the cheat sheet).
import hashlib
import json
import random
import re

from IPython.display import HTML, Markdown, display

# ---------------------------------------------------------------------------
# 🃏 The Wild Idea Deck
# ---------------------------------------------------------------------------
WILD_IDEA_DECK = [
    {"id": 1, "sector": "home",
     "idea": "An on-demand app that matches travelling plant owners with vetted 'plant sitters' who water, talk to, and send selfies of your plants."},
    {"id": 2, "sector": "entertainment",
     "idea": "An AI dream journal that turns parents' dreams into personalised bedtime stories for their kids."},
    {"id": 3, "sector": "pets",
     "idea": "Noise-cancelling headphones for dogs that switch on automatically during fireworks and thunderstorms."},
    {"id": 4, "sector": "food",
     "idea": "A smart lunchbox and companion app that turns eating vegetables into a game for picky kids."},
    {"id": 5, "sector": "education",
     "idea": "A language-learning app where you practise by debating AI versions of historical figures."},
    {"id": 6, "sector": "social",
     "idea": "'Grandparents-as-a-Service': matching retirees with busy families for storytelling, homework help and cooking lessons."},
    {"id": 7, "sector": "fitness",
     "idea": "Gym equipment that turns your workout into electricity and credits it to your home energy bill."},
    {"id": 8, "sector": "sustainability",
     "idea": "Carbon-negative coffee pods grown from mushroom mycelium that you can plant in your garden after use."},
    {"id": 9, "sector": "mobility",
     "idea": "A marketplace where office buildings rent their empty parking spots to food trucks at lunchtime."},
    {"id": 10, "sector": "entertainment",
     "idea": "A dating app for board-game lovers where every first date is a cooperative game night."},
]

# ---------------------------------------------------------------------------
# 📊 Fictional market data (numbers are made up for this exercise!)
# ---------------------------------------------------------------------------
MARKET_DATA = {
    "home": {"market_size_usd_bn": 48, "annual_growth_pct": 7.5, "avg_spend_usd_per_month": 25,
             "trends": ["Urban apartment gardening boom", "Gig-economy home services", "Smart-home sensors"],
             "risks": ["Trust & safety for in-home visits", "Seasonal demand (holidays)", "Low willingness to pay"],
             "benchmarks": {"typical_cac_usd": 30, "typical_monthly_churn_pct": 8}},
    "entertainment": {"market_size_usd_bn": 210, "annual_growth_pct": 9.2, "avg_spend_usd_per_month": 12,
                      "trends": ["Generative-AI personalised content", "Niche communities over mass platforms", "Offline-first experiences"],
                      "risks": ["Content moderation", "Subscription fatigue", "Fast-moving copycats"],
                      "benchmarks": {"typical_cac_usd": 18, "typical_monthly_churn_pct": 9}},
    "pets": {"market_size_usd_bn": 320, "annual_growth_pct": 6.1, "avg_spend_usd_per_month": 120,
             "trends": ["Pet humanisation ('fur babies')", "Premium wellness & anxiety products", "Subscription boxes"],
             "risks": ["Animal-welfare and vet claims", "Hardware returns", "Seasonal spikes (fireworks season)"],
             "benchmarks": {"typical_cac_usd": 35, "typical_monthly_churn_pct": 6}},
    "food": {"market_size_usd_bn": 95, "annual_growth_pct": 5.4, "avg_spend_usd_per_month": 40,
             "trends": ["Healthy eating for kids", "Gamified nutrition", "Connected kitchen devices"],
             "risks": ["Food-safety regulations", "Parents' screen-time concerns", "Hardware margins"],
             "benchmarks": {"typical_cac_usd": 40, "typical_monthly_churn_pct": 7}},
    "education": {"market_size_usd_bn": 165, "annual_growth_pct": 12.8, "avg_spend_usd_per_month": 15,
                  "trends": ["AI tutors and conversation practice", "Gamification", "Micro-learning"],
                  "risks": ["Low completion rates", "Historical accuracy & bias", "Free alternatives"],
                  "benchmarks": {"typical_cac_usd": 22, "typical_monthly_churn_pct": 10}},
    "social": {"market_size_usd_bn": 60, "annual_growth_pct": 8.0, "avg_spend_usd_per_month": 80,
               "trends": ["Loneliness economy", "Intergenerational connection", "Care marketplaces"],
               "risks": ["Safeguarding & background checks", "Two-sided marketplace cold start", "Insurance"],
               "benchmarks": {"typical_cac_usd": 55, "typical_monthly_churn_pct": 5}},
    "fitness": {"market_size_usd_bn": 105, "annual_growth_pct": 6.9, "avg_spend_usd_per_month": 45,
                "trends": ["Gamified workouts", "Sustainability-minded consumers", "Connected equipment"],
                "risks": ["Expensive hardware", "Energy regulations", "Motivation drop-off after January"],
                "benchmarks": {"typical_cac_usd": 60, "typical_monthly_churn_pct": 6}},
    "sustainability": {"market_size_usd_bn": 140, "annual_growth_pct": 14.5, "avg_spend_usd_per_month": 30,
                       "trends": ["Carbon-negative products", "Bio-materials (mycelium, algae)", "Circular economy"],
                       "risks": ["Greenwashing accusations", "Manufacturing scale-up", "Premium pricing"],
                       "benchmarks": {"typical_cac_usd": 45, "typical_monthly_churn_pct": 4}},
    "mobility": {"market_size_usd_bn": 75, "annual_growth_pct": 10.3, "avg_spend_usd_per_month": 200,
                 "trends": ["Under-used urban space", "Food-truck culture", "B2B marketplaces"],
                 "risks": ["City permits & zoning", "Liability", "Chicken-and-egg marketplace dynamics"],
                 "benchmarks": {"typical_cac_usd": 150, "typical_monthly_churn_pct": 3}},
}

# ---------------------------------------------------------------------------
# 🏢 Fictional competitor database
# ---------------------------------------------------------------------------
COMPETITORS = [
    {"name": "Leafy Nanny", "sector": "home", "tagline": "Plant-care tips from a chatbot", "keywords": ["plants", "plant care", "gardening"], "pricing": "$4.99/month", "weakness": "No in-person service"},
    {"name": "GreenThumb Robotics", "sector": "home", "tagline": "Self-watering smart pots", "keywords": ["plants", "smart pot", "watering", "iot"], "pricing": "$89 one-time", "weakness": "Only works with their own pots"},
    {"name": "SnoozeStory", "sector": "entertainment", "tagline": "AI bedtime stories on demand", "keywords": ["bedtime", "stories", "kids", "ai", "parents"], "pricing": "$7.99/month", "weakness": "Generic stories, nothing from the family's own life"},
    {"name": "DreamLog", "sector": "entertainment", "tagline": "Track and analyse your dreams", "keywords": ["dream", "journal", "sleep"], "pricing": "Free / $3.99 pro", "weakness": "Analysis only, no creative output"},
    {"name": "Roll for Love", "sector": "entertainment", "tagline": "Dating for video gamers", "keywords": ["dating", "games", "gamers"], "pricing": "Freemium, $14.99/month", "weakness": "Video games only, no tabletop community"},
    {"name": "StormHug", "sector": "pets", "tagline": "Anxiety-calming vests for dogs", "keywords": ["dog", "anxiety", "fireworks", "thunder"], "pricing": "$45 one-time", "weakness": "Doesn't address noise itself"},
    {"name": "Barkly Box", "sector": "pets", "tagline": "Monthly toy & treat box for dogs", "keywords": ["dog", "subscription", "toys", "treats"], "pricing": "$29/month", "weakness": "Commodity product, high churn"},
    {"name": "VeggieQuest", "sector": "food", "tagline": "Nutrition game app for kids", "keywords": ["kids", "vegetables", "game", "nutrition"], "pricing": "$5.99/month", "weakness": "Parents worry about extra screen time"},
    {"name": "LunchBotics", "sector": "food", "tagline": "Temperature-controlled smart bento", "keywords": ["lunchbox", "smart", "kids", "bento"], "pricing": "$129 one-time", "weakness": "Expensive, no engagement features"},
    {"name": "Lingo Legends", "sector": "education", "tagline": "Learn languages through quests", "keywords": ["language", "learning", "gamification"], "pricing": "Freemium, $9.99/month", "weakness": "Scripted dialogues, no real conversation"},
    {"name": "ChronoChat", "sector": "education", "tagline": "Chat with historical figures", "keywords": ["history", "historical figures", "chat", "ai"], "pricing": "$9.99/month", "weakness": "Trivia-focused, not designed for language practice"},
    {"name": "GrandBuddies", "sector": "social", "tagline": "Companion visits for seniors", "keywords": ["seniors", "retirees", "companionship", "elderly"], "pricing": "$25/hour", "weakness": "Treats seniors as recipients, not contributors"},
    {"name": "Village Nanny", "sector": "social", "tagline": "Babysitter marketplace", "keywords": ["childcare", "families", "babysitting", "marketplace"], "pricing": "15% booking fee", "weakness": "Transactional, no long-term relationships"},
    {"name": "PowerPedal", "sector": "fitness", "tagline": "Bike desks that charge your devices", "keywords": ["electricity", "energy", "bike", "workout"], "pricing": "$399 one-time", "weakness": "Tiny energy output, novelty factor"},
    {"name": "FitCoin Gym", "sector": "fitness", "tagline": "Gym that rewards workouts with points", "keywords": ["gym", "rewards", "workout", "points"], "pricing": "$35/month", "weakness": "Points have little real-world value"},
    {"name": "ShroomPack", "sector": "sustainability", "tagline": "Mycelium packaging for brands", "keywords": ["mycelium", "mushroom", "packaging", "b2b"], "pricing": "B2B contracts", "weakness": "No consumer brand"},
    {"name": "EcoPod Co.", "sector": "sustainability", "tagline": "Compostable coffee pods", "keywords": ["coffee", "pods", "compostable", "capsules"], "pricing": "$0.60/pod", "weakness": "Compostable is not carbon-negative"},
    {"name": "SpotSwap", "sector": "mobility", "tagline": "Peer-to-peer parking rental", "keywords": ["parking", "rental", "marketplace", "cars"], "pricing": "20% commission", "weakness": "Cars only, no commercial vehicles"},
    {"name": "TruckTrack", "sector": "mobility", "tagline": "Find food trucks near you", "keywords": ["food truck", "locator", "lunch"], "pricing": "Free (ads)", "weakness": "Consumer app only, doesn't solve where trucks can park"},
]

# ---------------------------------------------------------------------------
# 🦈 The Demo Day investor panel (fictional)
# ---------------------------------------------------------------------------
INVESTOR_PANEL = [
    {"name": "Ada Okafor", "nickname": "The Analyst", "fund": "Spreadsheet Capital",
     "loves": ["big markets", "healthy unit economics (LTV/CAC >= 3)", "defensible moats"],
     "hates": ["hand-wavy numbers", "tiny niches"],
     "catchphrase": "Show me the numbers."},
    {"name": "Rex Moon", "nickname": "The Visionary", "fund": "Moonshot Ventures",
     "loves": ["bold visions", "great storytelling", "creating new categories"],
     "hates": ["me-too products", "boring names"],
     "catchphrase": "Will this change how people live?"},
    {"name": "Priya Shah", "nickname": "The Operator", "fund": "Shipyard Partners",
     "loves": ["a tight MVP", "a clear go-to-market plan", "realistic milestones"],
     "hates": ["feature bloat", "vague target customers"],
     "catchphrase": "What ships in the first 90 days?"},
]

_TAKEN_NAMES = {"nova", "spark", "zen", "pulse", "flow", "bloom", "apex", "orbit", "nexus", "hive", "leaf", "dream"}


def draw_wild_idea(seed: int | None = None) -> dict:
    """Draw a random idea card from the Wild Idea Deck (pass a seed to make it repeatable)."""
    return random.Random(seed).choice(WILD_IDEA_DECK)


# ---------------------------------------------------------------------------
# 🔧 Tools (give these to your agents!)
# ---------------------------------------------------------------------------
def get_market_data(sector: str) -> str:
    """Get market size, growth, trends, risks and benchmark CAC/churn for a market sector.

    Args:
        sector: Market sector, one of: home, entertainment, pets, food, education, social, fitness, sustainability, mobility.

    Returns:
        JSON string with the (fictional) market data for the sector.
    """
    key = sector.strip().lower()
    if key not in MARKET_DATA:
        return json.dumps({"error": f"Unknown sector '{sector}'.", "available_sectors": sorted(MARKET_DATA)})
    return json.dumps({"sector": key, **MARKET_DATA[key]})


def search_competitors(keyword: str) -> str:
    """Search the competitor database for companies matching a keyword (e.g. 'dog', 'parking', 'coffee').

    Args:
        keyword: A single keyword or short phrase describing the product or customer.

    Returns:
        JSON string with up to 5 matching competitors, including their pricing and main weakness.
    """
    needle = keyword.strip().lower()
    matches = [
        c for c in COMPETITORS
        if needle and (needle in c["name"].lower() or needle in c["tagline"].lower()
                       or needle == c["sector"] or any(needle in k for k in c["keywords"]))
    ]
    if not matches:
        return json.dumps({"keyword": keyword, "competitors": [], "note": "No direct competitors found. Blue ocean, or no market?"})
    return json.dumps({"keyword": keyword, "competitors": matches[:5]})


def check_name_availability(name: str) -> str:
    """Check whether a startup name is available as a .com/.ai domain and whether it has trademark risk.

    Args:
        name: The proposed startup name, e.g. 'BarkBuds'.

    Returns:
        JSON string with domain availability, trademark risk and a verdict.
    """
    slug = re.sub(r"[^a-z0-9]", "", name.lower())
    if not slug:
        return json.dumps({"name": name, "error": "Name must contain letters or digits."})
    digest = int(hashlib.sha256(slug.encode()).hexdigest(), 16)
    clash = any(taken in slug for taken in _TAKEN_NAMES)
    com_free = not clash and digest % 3 != 0
    ai_free = not clash and digest % 4 != 0
    risk = "high" if clash else ("medium" if digest % 7 == 0 else "low")
    usable = (com_free or ai_free) and risk != "high"
    return json.dumps({
        "name": name,
        "domain_com": f"{slug}.com is " + ("available" if com_free else "taken"),
        "domain_ai": f"{slug}.ai is " + ("available" if ai_free else "taken"),
        "trademark_risk": risk,
        "verdict": "Looks usable!" if usable else "Pick another name.",
    })


def estimate_unit_economics(price_per_month_usd: float, cost_to_serve_per_month_usd: float,
                            acquisition_cost_usd: float, monthly_churn_pct: float) -> str:
    """Calculate customer lifetime value (LTV), LTV/CAC ratio and CAC payback period.

    For one-off purchases, spread the price over the expected product life to get a monthly figure.

    Args:
        price_per_month_usd: Average monthly revenue per customer in USD.
        cost_to_serve_per_month_usd: Average monthly cost to serve one customer in USD.
        acquisition_cost_usd: Customer acquisition cost (CAC) in USD.
        monthly_churn_pct: Percentage of customers lost each month, e.g. 5 for 5%.

    Returns:
        JSON string with gross margin, LTV, LTV/CAC, payback months and a verdict.
    """
    margin = price_per_month_usd - cost_to_serve_per_month_usd
    churn = max(monthly_churn_pct, 0.1) / 100
    lifetime_months = 1 / churn
    ltv = margin * lifetime_months
    ltv_cac = ltv / acquisition_cost_usd if acquisition_cost_usd > 0 else float("inf")
    payback = acquisition_cost_usd / margin if margin > 0 else float("inf")
    if margin <= 0:
        verdict = "Negative margin: you lose money on every customer."
    elif ltv_cac >= 3:
        verdict = "Healthy: investors like LTV/CAC >= 3."
    elif ltv_cac >= 1:
        verdict = "Risky: profitable per customer, but growth will be expensive."
    else:
        verdict = "Unsustainable: each customer costs more than they bring in."
    return json.dumps({
        "gross_margin_per_month_usd": round(margin, 2),
        "gross_margin_pct": round(100 * margin / price_per_month_usd, 1) if price_per_month_usd else None,
        "expected_lifetime_months": round(lifetime_months, 1),
        "ltv_usd": round(ltv, 2),
        "ltv_to_cac": round(ltv_cac, 2) if ltv_cac != float("inf") else "infinite",
        "cac_payback_months": round(payback, 1) if payback != float("inf") else "never",
        "verdict": verdict,
    })


def get_investor_panel() -> str:
    """Get the profiles of the Demo Day investors: what each one loves, hates and asks about.

    Returns:
        JSON string with the investor profiles.
    """
    return json.dumps(INVESTOR_PANEL)


# ---------------------------------------------------------------------------
# 📦 Launch Kit contract, checker and viewer
# ---------------------------------------------------------------------------
LAUNCH_KIT_SPEC = {
    "startup_name": "A catchy name you've checked is available",
    "one_liner": "A one-sentence elevator pitch",
    "target_customers": "Who exactly it's for (personas/segments)",
    "market_insights": "Market size, growth, trends and risks",
    "competitors": "Main competitors and how you'll beat them",
    "mvp_features": "A prioritised list of MVP features",
    "pricing": "Pricing model and price points",
    "unit_economics": "LTV, CAC, LTV/CAC and payback, calculated with the tool",
    "marketing_campaign": "Slogan, launch channels and a few social posts",
    "investor_pitch": "Your Demo Day pitch (e.g. a 6-slide outline or a 200-word script)",
    "investor_feedback": "What the investor panel said (scores, verdicts, what changed)",
}


def _to_plain(value):
    """Convert Pydantic models (and containers of them) into plain Python data."""
    if hasattr(value, "model_dump"):
        return _to_plain(value.model_dump())
    if isinstance(value, dict):
        return {k: _to_plain(v) for k, v in value.items()}
    if isinstance(value, (list, tuple)):
        return [_to_plain(v) for v in value]
    return value


def _is_filled(value) -> bool:
    if value is None:
        return False
    if isinstance(value, str):
        return bool(value.strip())
    if isinstance(value, (list, tuple, dict, set)):
        return len(value) > 0
    return bool(str(value).strip())


def check_launch_kit(kit) -> bool:
    """Mission control check: verify that the launch kit contains every required section."""
    kit = _to_plain(kit)
    if not isinstance(kit, dict):
        print(f"❌ launch_kit must be a dict (or a Pydantic model), got {type(kit).__name__}.")
        return False
    passed = 0
    for key, description in LAUNCH_KIT_SPEC.items():
        ok = _is_filled(kit.get(key))
        passed += ok
        print(f"{'✅' if ok else '❌'} {key:<20}" + ("" if ok else f" <- missing: {description}"))
    extras = [k for k in kit if k not in LAUNCH_KIT_SPEC]
    if extras:
        print(f"\n✨ Extra sections (nice!): {', '.join(extras)}")
    print(f"\nLaunch readiness: {passed}/{len(LAUNCH_KIT_SPEC)}")
    if passed == len(LAUNCH_KIT_SPEC):
        print("🚀 All systems go! Your startup is ready for Demo Day.")
    return passed == len(LAUNCH_KIT_SPEC)


def _render(value, indent: int = 0) -> str:
    pad = "  " * indent
    if isinstance(value, dict):
        lines = []
        for k, v in value.items():
            if isinstance(v, (dict, list)):
                lines.append(f"{pad}- **{k}**:\n{_render(v, indent + 1)}")
            else:
                lines.append(f"{pad}- **{k}**: {v}")
        return "\n".join(lines)
    if isinstance(value, list):
        return "\n".join(
            f"{pad}-\n{_render(v, indent + 1)}" if isinstance(v, (dict, list)) else f"{pad}- {v}" for v in value
        )
    return f"{pad}{value}"


def show_launch_kit(kit) -> None:
    """Render the launch kit as a nicely formatted Markdown report."""
    kit = _to_plain(kit)
    if not isinstance(kit, dict) or not kit:
        print("Nothing to show yet. Build your launch_kit first!")
        return
    md = [f"# 🚀 {kit.get('startup_name', 'Unnamed Startup')}"]
    if kit.get("one_liner"):
        md.append(f"> {kit['one_liner']}")
    for key, value in kit.items():
        if key in ("startup_name", "one_liner", "landing_page_html") or not _is_filled(value):
            continue
        md.append(f"## {key.replace('_', ' ').title()}\n\n{_render(value)}")
    display(Markdown("\n\n".join(md)))
    if kit.get("landing_page_html"):
        display(HTML(kit["landing_page_html"]))


print(f"🧰 Startup Toolkit loaded: {len(WILD_IDEA_DECK)} wild ideas, {len(MARKET_DATA)} sectors, "
      f"{len(COMPETITORS)} competitors, {len(INVESTOR_PANEL)} investors.")
````

**Framework:** LangGraph. My workflow really is a graph: two branches that don't depend on each other at the start, a join, and a loop that must stop. **Patterns I use (where and why)**
- **⚡ Parallelization:** market research and naming don't need each other, so both start from `START` at the same time and meet again at a join (`["product", "brand"] → campaign`). The 3 investors also score in parallel with `batch`, so none of them sees the others' scores and copies them.
- **🔗 Prompt chaining:** market → product → campaign → pitch. Every agent returns a Pydantic model, and the next one gets a typed `studio_brief`, not a chat history. So the numbers in the pitch are the same numbers the calculator produced.
- **🔁 Evaluator-optimizer (Shark Tank loop):** the investors score, and the Pitch Coach revises from their "must fix" points. A pitch is subjective, and a separate judge with a different goal (Ada wants numbers, Rex a vision, Priya a 90-day plan) catches things the writer doesn't.
- **Also used:** tool use / ReAct inside the agents, and human-in-the-loop on the name.

**Unit economics are taken from the tool, not from the LLM.** A callback (`StageMeter`) logs every tool call with its inputs and its raw result. `launch_kit["unit_economics"]` is the last `estimate_unit_economics` result from that log, so the numbers can't be changed by the model retelling them.

**Stop conditions and safety limits**
- Shark Tank: stops at an average score ≥ 8, **or** after 3 rounds, **or** once the run has used 45 LLM calls (the challenge limit is about 50).
- Every ReAct agent runs with `recursion_limit=14`, so a ReAct loop can't run away.
- Brand Studio: at most 5 name checks. CFO: at most 3 calculator runs. These two are rules in the prompts, and `recursion_limit` is the hard cap behind them. Founder: at most 1 veto (enforced in code).
- One run is about 25–26 LLM calls and about 1 minute (measured, see the stats table in section 6).

**Tips from the Moonshot engineering team:**
- 🐣 **Start small:** get one agent with one tool working end-to-end, then add agents one by one.
- 🔎 **Print intermediate results.** Debugging multi-agent systems is mostly reading what the agents said.
- 🎯 **Write specific prompts:** give each agent a role, its inputs, and the expected output format and length.
- 🧮 **Let tools do the math, and let LLMs do the judgement and creativity.**
- 🛑 **Always cap loops** with `max_steps`, a maximum number of rounds, or `max_iterations`.
- 🌡️ **GPT-5 models** only accept the default `temperature` and reject `max_tokens` and `stop`, so don't set them.
- ⏳ **Async in notebooks:** you can `await` directly in a cell (LlamaIndex, ADK, CrewAI's `kickoff_async`).
- 📐 **Structured outputs** (Pydantic models) make assembling the `launch_kit` much easier.

````python
# 5.1 Model, tools and a small meter that counts LLM calls, tool calls, tokens and seconds per stage
import time
from collections import defaultdict
from typing import Literal, TypedDict

from langchain.agents import create_agent
from langchain_core.callbacks import BaseCallbackHandler
from langchain_core.tools import tool
from langchain_openai import ChatOpenAI
from langgraph.graph import END, START, StateGraph
from pydantic import BaseModel, Field

llm = ChatOpenAI(model=AZURE_OPENAI_DEPLOYMENT, base_url=AZURE_OPENAI_ENDPOINT,
                 api_key=os.environ["AZURE_OPENAI_API_KEY"])

# the toolkit functions already have type hints + docstrings, so tool() is all they need
market_tool = tool(get_market_data)
competitor_tool = tool(search_competitors)
name_tool = tool(check_name_availability)
economics_tool = tool(estimate_unit_economics)
panel_tool = tool(get_investor_panel)

STATS = defaultdict(lambda: {"llm_calls": 0, "tool_calls": 0, "tokens": 0, "seconds": 0.0})
TOOL_LOG = []  # every tool call an agent made: stage, tool, args, raw result


class StageMeter(BaseCallbackHandler):
    """Callback for one stage: counts model calls, tokens and tool calls, and keeps the tool results."""

    def __init__(self, stage):
        self.stage = stage
        self.open_tools = {}

    def on_llm_end(self, response, **kwargs):
        STATS[self.stage]["llm_calls"] += 1
        message = getattr(response.generations[0][0], "message", None)
        usage = getattr(message, "usage_metadata", None) or {}
        STATS[self.stage]["tokens"] += usage.get("total_tokens", 0)

    def on_tool_start(self, serialized, input_str, *, run_id, inputs=None, **kwargs):
        self.open_tools[run_id] = (serialized.get("name"), inputs or input_str)

    def on_tool_end(self, output, *, run_id, **kwargs):
        name, args = self.open_tools.pop(run_id, ("?", None))
        STATS[self.stage]["tool_calls"] += 1
        TOOL_LOG.append({"stage": self.stage, "tool": name, "args": args,
                         "result": getattr(output, "content", output)})


def metered(stage, runnable, payload):
    """Run an agent or a model under a StageMeter and add the time to that stage."""
    start = time.time()
    out = runnable.invoke(payload, config={"callbacks": [StageMeter(stage)], "recursion_limit": 14})
    STATS[stage]["seconds"] += time.time() - start
    return out


def ask(text):
    return {"messages": [{"role": "user", "content": text}]}


print("🔧 Tools ready:", ", ".join(t.name for t in [market_tool, competitor_tool, name_tool, economics_tool, panel_tool]))
````

````python
# 5.2 The team. Every agent answers with a Pydantic model, so the launch kit is assembled from typed fields.

class Competitor(BaseModel):
    name: str
    their_pricing: str
    their_weakness: str
    how_we_win: str


class MarketReport(BaseModel):
    target_customers: list[str] = Field(description="2-3 concrete personas: 'Name, age, situation: what they need'")
    market_size: str
    growth: str
    trends: list[str]
    risks: list[str] = Field(description="the sector risks, each one applied to THIS idea")
    competitors: list[Competitor]
    benchmark_cac_usd: float
    benchmark_monthly_churn_pct: float


class Brand(BaseModel):
    startup_name: str
    one_liner: str
    why_this_name: str
    names_rejected: list[str] = Field(description="names you checked that were taken or risky")


class PriceTier(BaseModel):
    tier: str
    price: str
    includes: str


class ProductPlan(BaseModel):
    mvp_features: list[str] = Field(description="max 6, ordered by priority, each starting with P1/P2/P3")
    pricing_model: str
    tiers: list[PriceTier] = Field(description="real price tiers only; a single tier is fine, no notes as tiers")
    price_per_month_usd: float
    cost_to_serve_per_month_usd: float
    cac_usd: float
    monthly_churn_pct: float
    assumptions: str = Field(description="where each number you put into the calculator comes from")


class Campaign(BaseModel):
    slogan: str
    launch_channels: list[str]
    social_posts: list[str] = Field(description="3 posts, each starting with its platform, e.g. 'TikTok: ...'")
    launch_stunt: str


class Slide(BaseModel):
    title: str
    content: str


class Pitch(BaseModel):
    slides: list[Slide] = Field(description="exactly 6 slides")
    closing_line: str
    changes_made: list[str] = Field(description="what changed since the last version and which investor asked for it; empty list for version 1")


class Verdict(BaseModel):
    score: int = Field(description="1-10")
    decision: Literal["invest", "maybe", "pass"]
    liked: str
    concern: str
    must_fix: str = Field(description="one concrete change that would raise your score")


# 🔎 Market Scout (ReAct, 2 tools)
scout = create_agent(
    llm, tools=[market_tool, competitor_tool], response_format=MarketReport, name="market_scout",
    system_prompt=(
        "You are the Market Scout of a venture studio. Call get_market_data for the sector, and call "
        "search_competitors with 2-3 different keywords that fit the idea (customers, the job it does, the "
        "sector). If those find nothing, also search the sector name itself to find adjacent players. Every number "
        "you report must come from a tool result. Never list 'no competitor' as a competitor: list the real ones "
        "you found plus the status quo customers use today (e.g. doing it themselves), marked as status quo. "
        "For every competitor, say in one line how this idea wins against it."))

# 🏷️ Brand Studio (ReAct, keeps checking until a name is free)
brand_studio = create_agent(
    llm, tools=[name_tool], response_format=Brand, name="brand_studio",
    system_prompt=(
        "You are a naming director. Brainstorm short, playful, easy-to-say names. Check names with "
        "check_name_availability ONE AT A TIME, and stop at the first one whose verdict is 'Looks usable!' and "
        "whose trademark_risk is 'low'. Check at most 5 names. Never return a name you did not check. "
        "Then write a one-sentence elevator pitch (max 25 words)."))

# 🧮 Product & CFO (ReAct, the calculator decides the price)
product_lead = create_agent(
    llm, tools=[economics_tool, market_tool], response_format=ProductPlan, name="product_cfo",
    system_prompt=(
        "You are the product lead and CFO in one. Plan an MVP that can ship in 90 days (no feature bloat) and "
        "a price. Build the monthly cost to serve one customer line by line (people's time, transport, materials, "
        "insurance, software). Start from the sector's benchmark CAC and churn; a product priced far above the "
        "sector's average spend is harder to sell, so its CAC should be higher too. Then call "
        "estimate_unit_economics. The believable range is LTV/CAC between 3 and 8: below 3 the business doesn't "
        "work, above 8 investors assume the costs are fantasy. If the result is outside that range, fix the "
        "input that is least realistic and call the calculator again, at most 3 calls in total. The numbers in "
        "your answer must be exactly the inputs of your LAST calculator call, and the assumptions must show the "
        "cost lines and why each number changed between calls."))

# 📣 Campaign Lead (one structured call, no tools needed)
campaign_lead = llm.with_structured_output(Campaign)
CAMPAIGN_PROMPT = ("You are a scrappy launch marketer. Write a campaign for a startup with almost no budget: a "
                   "slogan under 8 words, 3-4 launch channels that fit the target customers, 3 social posts, and "
                   "one memorable launch stunt. Funny is good, cringe is not.")

# 🎤 Pitch Coach (ReAct, reads the investor panel before writing)
pitch_coach = create_agent(
    llm, tools=[panel_tool], response_format=Pitch, name="pitch_coach",
    system_prompt=(
        "You write Demo Day pitches: exactly 6 slides (problem, solution, market, business model with unit "
        "economics, go-to-market and 90-day plan, team ask). Check the investor panel with get_investor_panel "
        "before the first version, so the pitch speaks to what they love. Use ONLY the facts and numbers in "
        "the studio brief. If an investor wants data we don't have (traction, partners), never invent it: turn "
        "it into a concrete test or milestone instead."))

# 🦈 The investor panel: each investor is its own agent with its own persona
investor_judge = llm.with_structured_output(Verdict)


def investor_prompt(inv):
    return (f"You are {inv['name']}, '{inv['nickname']}' from {inv['fund']}, judging Demo Day pitches. "
            f"You love: {', '.join(inv['loves'])}. You hate: {', '.join(inv['hates'])}. "
            f"Your catchphrase: '{inv['catchphrase']}'. Score the pitch 1-10 from YOUR point of view. Be tough: "
            "8 or more means you would sign a cheque today, and most first pitches land at 5-7. Judge only what "
            "is written in the pitch, and give one concrete fix.")


# 🌐 Web Designer (one call, plain HTML back)
WEB_PROMPT = ("You build landing pages. Return ONE self-contained HTML document: inline CSS only, no JavaScript, "
              "no external images, fonts or links. Sections: hero with name, slogan and a sign-up button, "
              "3 features, pricing tiers, a short FAQ. Return only the HTML, no explanations.")

print("🤖 Team hired: market_scout, brand_studio, product_cfo, campaign_lead, pitch_coach, 3 investors, web_designer")
````

````python
# 5.3 The workflow: parallel start -> chain -> Shark Tank loop -> landing page

FOUNDER_IN_THE_LOOP = False  # True: approve or veto the name with input(). False: Run All never waits for you.
TARGET_SCORE = 8             # Shark Tank stops when the average investor score reaches this...
MAX_ROUNDS = 3               # ...or after this many rounds...
CALL_BUDGET = 45             # ...or when the run has used this many LLM calls


class StudioState(TypedDict, total=False):
    idea: str
    sector: str
    market: MarketReport
    brand: Brand
    founder_note: str
    product: ProductPlan
    campaign: Campaign
    pitch: Pitch
    rounds: list
    landing_page_html: str


def tool_unit_economics():
    """The real calculator output (from the tool log), so the kit never relies on the LLM retelling numbers."""
    runs = [t for t in TOOL_LOG if t["tool"] == "estimate_unit_economics"]
    if not runs:
        return {"warning": "the product agent never called the calculator"}
    last = runs[-1]
    return {"inputs": last["args"], **json.loads(last["result"]), "calculator_runs": len(runs)}


def studio_brief(state):
    """Everything the studio knows so far, as one text block for the later agents."""
    lines = [f"Idea: {state['idea']}"]
    if "brand" in state:
        lines.append(f"Name: {state['brand'].startup_name}. One-liner: {state['brand'].one_liner}")
    if "market" in state:
        m = state["market"]
        lines += [f"Target customers: {m.target_customers}",
                  f"Market: {m.market_size}, growth {m.growth}. Trends: {m.trends}. Risks: {m.risks}",
                  "Competitors: " + "; ".join(f"{c.name} ({c.their_pricing}) - we win because {c.how_we_win}"
                                               for c in m.competitors)]
    if "product" in state:
        p = state["product"]
        lines += [f"MVP features: {p.mvp_features}",
                  f"Pricing: {p.pricing_model}. Tiers: {[t.model_dump() for t in p.tiers]}",
                  f"Unit economics from the calculator: {json.dumps(tool_unit_economics())}",
                  f"Cost lines and assumptions behind those numbers: {p.assumptions}"]
    if "campaign" in state:
        lines.append(f"Campaign: {state['campaign'].model_dump_json()}")
    return "\n".join(lines)


def pitch_as_text(pitch):
    slides = "\n".join(f"Slide {i}: {s.title}\n{s.content}" for i, s in enumerate(pitch.slides, 1))
    return f"{slides}\nClosing line: {pitch.closing_line}"


# ---- nodes ----
def market_node(state):
    report = metered("market", scout, ask(f"Idea: {state['idea']}\nSector: {state['sector']}"))
    print(f"🔎 Market done: {len(report['structured_response'].competitors)} competitors found")
    return {"market": report["structured_response"]}


def brand_node(state):
    request = f"Idea: {state['idea']}\nFind an available name and write the one-liner."
    brand = metered("brand", brand_studio, ask(request))["structured_response"]
    note = "auto-approved (FOUNDER_IN_THE_LOOP is off)"
    if FOUNDER_IN_THE_LOOP:
        answer = input(f"🧑‍💼 Founder check: name '{brand.startup_name}'. Enter = approve, 'veto' = new name: ")
        if answer.strip().lower() == "veto":
            vetoed = brand.startup_name
            brand = metered("brand", brand_studio, ask(request + f"\nThe founder vetoed '{vetoed}'. Pick another name."))["structured_response"]
            note = f"founder vetoed '{vetoed}', took the second pick"
        else:
            note = "approved by the founder"
    print(f"🏷️ Name: {brand.startup_name} ({note})")
    return {"brand": brand, "founder_note": note}


def product_node(state):
    m = state["market"]
    brief = (f"Idea: {state['idea']}\nSector: {state['sector']}\nTarget customers: {m.target_customers}\n"
             f"Competitors: {[c.model_dump() for c in m.competitors]}\n"
             f"Benchmarks: CAC ${m.benchmark_cac_usd}, churn {m.benchmark_monthly_churn_pct}% per month")
    plan = metered("product", product_lead, ask(brief))["structured_response"]
    print(f"🧮 Product done: {len(plan.mvp_features)} MVP features, LTV/CAC = {tool_unit_economics().get('ltv_to_cac')}")
    return {"product": plan}


def campaign_node(state):
    campaign = metered("campaign", campaign_lead, [("system", CAMPAIGN_PROMPT), ("user", studio_brief(state))])
    print(f"📣 Slogan: {campaign.slogan}")
    return {"campaign": campaign}


def pitch_node(state):
    rounds = state.get("rounds", [])
    if not rounds:
        request = studio_brief(state) + "\n\nWrite version 1 of the pitch."
    else:
        feedback = "\n".join(f"- {v['investor']} ({v['score']}/10): concern: {v['concern']} | must fix: {v['must_fix']}"
                             for v in rounds[-1]["verdicts"])
        request = (studio_brief(state) + f"\n\nCurrent pitch:\n{pitch_as_text(state['pitch'])}\n\n"
                   f"Investor feedback from round {len(rounds)}:\n{feedback}\n\n"
                   f"Write version {len(rounds) + 1}: fix every concern and list the changes in changes_made.")
    pitch = metered(f"pitch_v{len(rounds) + 1}", pitch_coach, ask(request))["structured_response"]
    print(f"🎤 Pitch v{len(rounds) + 1} written ({len(pitch.slides)} slides)")
    return {"pitch": pitch}


def investors_node(state):
    n = len(state.get("rounds", [])) + 1
    name = state["brand"].startup_name
    last = {v["investor"]: v for v in state["rounds"][-1]["verdicts"]} if state.get("rounds") else {}
    chats = []
    for inv in INVESTOR_PANEL:
        message = f"Pitch for {name} (version {n}):\n\n{pitch_as_text(state['pitch'])}"
        if inv["name"] in last:  # investors remember what they asked for last round
            before = last[inv["name"]]
            message += (f"\n\nLast round you gave {before['score']}/10 and asked: {before['must_fix']}\n"
                        "Check whether this version really fixed it. Raise the score only for real progress, "
                        "and lower it if the pitch got worse.")
        chats.append([("system", investor_prompt(inv)), ("user", message)])
    start = time.time()
    verdicts = investor_judge.batch(chats, config={"callbacks": [StageMeter(f"investors_r{n}")]})  # in parallel
    STATS[f"investors_r{n}"]["seconds"] += time.time() - start
    rows = [{"investor": inv["name"], **v.model_dump()} for inv, v in zip(INVESTOR_PANEL, verdicts)]
    average = round(sum(r["score"] for r in rows) / len(rows), 2)
    print(f"🦈 Round {n}: " + ", ".join(f"{r['investor']} {r['score']}" for r in rows) + f" -> average {average}")
    entry = {"round": n, "average_score": average, "verdicts": rows, "changes_in_this_pitch": state["pitch"].changes_made}
    return {"rounds": state.get("rounds", []) + [entry]}


def web_node(state):
    html = metered("web", llm, [("system", WEB_PROMPT), ("user", studio_brief(state))]).content.strip()
    html = html.removeprefix("```html").removeprefix("```").removesuffix("```").strip()
    print(f"🌐 Landing page: {len(html):,} characters of HTML")
    return {"landing_page_html": html}


def total_llm_calls():
    return sum(s["llm_calls"] for s in STATS.values())


def stop_reason(rounds):
    if rounds[-1]["average_score"] >= TARGET_SCORE:
        return f"average score {rounds[-1]['average_score']} reached the target of {TARGET_SCORE}"
    if len(rounds) >= MAX_ROUNDS:
        return f"hit the limit of {MAX_ROUNDS} rounds"
    if total_llm_calls() >= CALL_BUDGET:
        return f"hit the budget of {CALL_BUDGET} LLM calls"
    return None


def after_investors(state):
    return "web" if stop_reason(state["rounds"]) else "pitch"


# ---- graph ----
graph = StateGraph(StudioState)
for node_name, fn in [("market", market_node), ("brand", brand_node), ("product", product_node),
                      ("campaign", campaign_node), ("pitch", pitch_node), ("investors", investors_node), ("web", web_node)]:
    graph.add_node(node_name, fn)

graph.add_edge(START, "market")               # parallel start: research and naming don't need each other
graph.add_edge(START, "brand")
graph.add_edge("market", "product")           # chain: product uses the market report
graph.add_edge(["product", "brand"], "campaign")  # join: wait for both branches
graph.add_edge("campaign", "pitch")
graph.add_edge("pitch", "investors")
graph.add_conditional_edges("investors", after_investors, {"pitch": "pitch", "web": "web"})  # Shark Tank loop
graph.add_edge("web", END)

studio = graph.compile()
print(studio.get_graph().draw_mermaid())
````

## 6. 🚀 Launch!

Run your venture studio on `MY_IDEA` and collect the results into the `launch_kit` dictionary. You can fill it straight from your agents' outputs, or have a final "assembler" agent build it.

````python
# Run the studio on MY_IDEA and build the launch kit
STATS.clear()
TOOL_LOG.clear()
run_start = time.time()
final = studio.invoke({"idea": MY_IDEA["idea"], "sector": MY_IDEA["sector"]})
run_seconds = time.time() - run_start

market, brand, product, campaign, pitch = (final[k] for k in ["market", "brand", "product", "campaign", "pitch"])
rounds = final["rounds"]
name_checks = [json.loads(t["result"]) for t in TOOL_LOG if t["tool"] == "check_name_availability"]
chosen_check = next((c for c in reversed(name_checks) if c["name"].lower() == brand.startup_name.lower()), None)

launch_kit = {
    "startup_name": brand.startup_name,
    "one_liner": brand.one_liner,
    "target_customers": market.target_customers,
    "market_insights": {"market_size": market.market_size, "growth": market.growth,
                        "trends": market.trends, "risks": market.risks},
    "competitors": [c.model_dump() for c in market.competitors],
    "mvp_features": product.mvp_features,
    "pricing": {"model": product.pricing_model, "tiers": [t.model_dump() for t in product.tiers]},
    "unit_economics": {**tool_unit_economics(), "assumptions": product.assumptions},
    "marketing_campaign": campaign.model_dump(),
    "investor_pitch": {"slides": [f"{s.title}: {s.content}" for s in pitch.slides], "closing_line": pitch.closing_line},
    "investor_feedback": {
        "score_path": [r["average_score"] for r in rounds],
        "stopped_because": stop_reason(rounds),
        "rounds": rounds,
    },
    "name_check": {"why_this_name": brand.why_this_name, "names_rejected": brand.names_rejected,
                   "tool_result_for_chosen_name": chosen_check, "founder": final["founder_note"]},
    "run_stats": {
        "llm_calls": total_llm_calls(),
        "tool_calls": sum(s["tool_calls"] for s in STATS.values()),
        "tokens": sum(s["tokens"] for s in STATS.values()),
        "wall_seconds": round(run_seconds, 1),
    },
    "landing_page_html": final["landing_page_html"],
}

print(f"\n{'stage':<14}{'LLM calls':>10}{'tools':>7}{'tokens':>9}{'seconds':>9}")
for stage, s in STATS.items():
    print(f"{stage:<14}{s['llm_calls']:>10}{s['tool_calls']:>7}{s['tokens']:>9,}{s['seconds']:>9.1f}")
print(f"{'TOTAL':<14}{total_llm_calls():>10}{launch_kit['run_stats']['tool_calls']:>7}"
      f"{launch_kit['run_stats']['tokens']:>9,}{run_seconds:>9.1f}  (wall clock, parallel stages overlap)")
print(f"\n🦈 Score path: {launch_kit['investor_feedback']['score_path']}, stopped because {stop_reason(rounds)}")
print(f"🏷️ Name check for '{brand.startup_name}': {chosen_check['verdict'] if chosen_check else 'NOT CHECKED!'}")
````

## 8. 🪞 Reflection

I ran the whole thing 5 times while building it. In runs 1–4 I fixed one thing each time, and run 5 only added the cell that saves `landing_page.html` and `launch_kit.json`. The outputs you see above are from run 5.

| Run | What I changed before it | LLM calls | Time | LTV/CAC | Investor averages |
|---|---|---|---|---|---|
| 1 | first version | 25 | 55.7 s | 16.25 | 6.0 → 7.0 → 6.67 |
| 2 | competitor fix, CFO costs line by line, investors remember their last verdict | 26 | 61.0 s | 37.92 😬 | 6.33 → 7.0 → 7.33 |
| 3 | hard LTV/CAC range of 3–8 for the CFO | 25 | 53.7 s | 4.09 | 6.0 → 7.33 → 7.0 |
| 4 | Pitch Coach also gets the CFO's cost lines, no filler price tiers | 26 | 66.3 s | 8.0 | 6.0 → 7.0 → 7.67 |
| 5 | only the save-to-files cell (agents unchanged) | 25 | 55.5 s | 4.17 | 6.33 → 7.0 → 7.0 |

**1. Which pattern added the most value, and why?**

The investor loop, easily. In run 5 (the one above) the first pitch got 6/7/6, and honestly all three investors said the same thing: cute idea, but where is the proof it's a real business? What I liked is that each of them asked for something different. Ada said "$3 insurance and $8 transport feel hand-wavy", Rex wanted 10 paying customers on one route with real costs and service time, and Priya wanted one beachhead, like 20 homeowners in one neighbourhood. By version 3 the pitch had one pilot neighbourhood, Nextdoor as the *one* main channel instead of five, 20 paying customers at $49/month, a list of what the pilot has to measure, and the launch stunt moved down to "waitlist generator". I wouldn't have written that plan myself on the first try. In run 4 the same loop got Rex and Priya to "invest". The parallel start saved a few seconds, but the loop is what actually changed the content.

**2. Where did the agents struggle, and how did I fix it?**

Quite a lot, actually:

- **Competitors.** In run 1 the Scout searched only for goat stuff, found nothing, and then listed "No direct competitors found" three times *as competitors*. I told it to also search the sector name and to add the status quo. Now there's Leafy Nanny from the database, DIY lawn care (the real competitor anyway) and normal lawn-mowing companies. For those the Scout honestly wrote "Not provided by tool data" instead of making up a price.
- **Unrealistic numbers.** Run 1 had LTV/CAC 16 with $40/month to serve a customer with goats, a handler and a van, which is way too cheap. So I wrote "above 10 looks like fantasy" in the prompt, and it got worse 😅: run 2 put the price up to $149, kept CAC at $30 and got 37.9. A soft hint just didn't work. A hard rule did ("the believable range is 3–8, if you're outside, fix the worst input and call the calculator again"). In run 4 it really used the retry: the first call gave 8.75, so it raised CAC from $60 to $75 and called again. But it also changed the price and the cost at the same time, and it landed on *exactly* 8.0, the edge of my range. So it was aiming at my rule more than at reality, which is a bit funny but also a real problem.
- **Investors with no memory.** In run 1 the score went down in the last round (7.0 → 6.67), because every round they judged from zero. I gave each investor their own last score and "must fix", and after that they actually check ("You addressed my last concern directly"). It still doesn't fully stop them from moving the goalposts. In run 3 Priya wrote "You did what I asked" and still went from 8 to 7, because she found new things to want. And in run 5 the pitch's own change list says "Tightened the MVP", while Priya in the same round says the MVP "is starting to creep again", so the writer and the judge don't even agree on what changed.
- **My own bug in the chain.** In run 3 Ada asked to see the $21 cost broken down, and the pitch could only list the categories without the amounts. The reason was my `studio_brief` passed the calculator result to the Pitch Coach but not the CFO's cost lines. It simply didn't have the numbers. One extra line fixed it, and since run 4 the business-model slide shows the cost lines (run 5: $14 + $8 + $7 + $3 + $2 = $34). A side effect I didn't expect: once the numbers were visible, the feedback got sharper. In run 5 Ada went straight for "$3 insurance and $8 transport feel hand-wavy". Good reminder that an agent only knows what you give it, and the judges too.
- **A fake price tier.** The schema wanted a *list* of tiers, there was only one plan, so the CFO made a "tier" called "Note" with the price "No higher tiers in MVP". One field description fixed it.
